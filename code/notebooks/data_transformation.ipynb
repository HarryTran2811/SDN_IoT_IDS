```python
import os
import numpy as np
from PIL import Image

def save_images_from_df(df, folder_name):
    count = 0
    ims = []

    image_path = f"/home/cc1/Documents/Code/AIO-IDS/data/image_cicids2017/train/{folder_name}/"
    os.makedirs(image_path, exist_ok=True)

    for i in range(len(df)):  
        count += 1
        if count <= 27: 
            im = df.iloc[i].values
            ims = np.append(ims, im)
        else:
            ims = np.array(ims).reshape(9, 9, 3)
            array = np.array(ims, dtype=np.uint8)
            new_image = Image.fromarray(array)
            new_image.save(image_path + str(i) + '.png')
            count = 0
            ims = []

for idx in range(15):  
    df_name = f"df{idx}" 
    
    df = globals().get(df_name)
    
    if df is not None:  # Kiểm tra xem DataFrame có tồn tại không
        folder_name = str(idx)  # Tạo tên thư mục tương ứng
        save_images_from_df(df, folder_name)  # Gọi hàm lưu ảnh
    else:
        print(f"DataFrame {df_name} không tồn tại.")
```

```python
import cv2
def get_224(folder,dstdir):
    imgfilepaths=[]
    for root,dirs,imgs in os.walk(folder):
        for thisimg in imgs:
            thisimg_path=os.path.join(root,thisimg)
            imgfilepaths.append(thisimg_path)
    for thisimg_path in imgfilepaths:
        dir_name,filename=os.path.split(thisimg_path)
        dir_name=dir_name.replace(folder,dstdir)
        new_file_path=os.path.join(dir_name,filename)
        if not os.path.exists(dir_name):
            os.makedirs(dir_name)
        img=cv2.imread(thisimg_path)
        img=cv2.resize(img,(224,224))
        cv2.imwrite(new_file_path,img)
    print('Finish resizing'.format(folder=folder))
```

```python
DATA_DIR_224='/home/cc1/Documents/Code/AIO-IDS/data/image_cicids2017/train_224'
DATA_DIR_9 = '/home/cc1/Documents/Code/AIO-IDS/data/image_cicids2017/train'
get_224(folder=DATA_DIR_9,dstdir=DATA_DIR_224)
```
