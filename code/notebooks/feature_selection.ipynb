```python
import pandas as pd

train_df= pd.read_csv('/home/cc1/Documents/Code/AIO-IDS/data/out_cicids2017/train_MachineLearningCVE.csv')
```

```python
train_df=train_df.drop(columns=['Unnamed: 0'])
X = train_df.drop(columns=['Label'])
```

```python
y = train_df['Label']
```

```python
from sklearn.feature_selection import SequentialFeatureSelector
from sklearn.ensemble import RandomForestClassifier

model = RandomForestClassifier(n_estimators=100)
sfs = SequentialFeatureSelector(model, n_features_to_select=9)
sfs.fit(X, y)

X_sfs = sfs.transform(X)
print("Selected features:", sfs.get_support())
```
