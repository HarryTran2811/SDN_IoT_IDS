```python
import os

import pandas as pd
import numpy as np

from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
```

```python
PROCESSED_DIR_PATH = "/home/cc1/Documents/Code/AIO-IDS/data/out_cicids2017"
FILE_PATH = '/home/cc1/Documents/Code/AIO-IDS/data/cicids2017.csv'
```

```python
def label_encoding() -> LabelEncoder:
    le = LabelEncoder()

    labels = pd.read_csv(FILE_PATH, usecols=['Label'], skipinitialspace=True)

    le.fit(labels.Label)

    np.save(os.path.join(PROCESSED_DIR_PATH, 'label_encoder.npy'), le.classes_)

    print("Total rows: {}".format(labels.shape))
    print("Class distribution:\n{}\n".format(labels.Label.value_counts()))

    return le
```

```python
def process(df: pd.DataFrame, le: LabelEncoder) -> (np.ndarray, np.ndarray):
    df.Label = le.transform(df.Label)
    nan_rows = df[df.isna().any(axis=1)].shape[0]
    print("Fill NaN in {} rows with average value of each class.".format(nan_rows))
    df.iloc[:, df.columns != "Label"] = df.groupby("Label").transform(lambda x: x.fillna(x.mean()))

    inf_rows = df[df.isin([np.inf]).any(axis=1)].shape[0]
    print("Replace Inf in {} rows with maximum value of each class.".format(inf_rows))
    df = df.replace([np.inf], np.nan)
    df.iloc[:, df.columns != "Label"] = df.groupby("Label").transform(lambda x: x.fillna(x.max()))

    print("Replace negative values with minimum value of each class.")
    df[df < 0] = np.nan
    df.iloc[:, df.columns != "Label"] = df.groupby("Label").transform(lambda x: x.fillna(x.min()))

    return df
```

```python
def split_train_test(df: pd.DataFrame) -> (pd.DataFrame, pd.DataFrame):
    x = df.iloc[:, df.columns != 'Label']
    y = df[['Label']]

    x_train, x_test, y_train, y_test = train_test_split(x, y, stratify=y, test_size=0.20,
                                                        random_state=np.random.randint(10))

    del x, y

    train = pd.concat([x_train, y_train], axis=1, sort=False)
    test = pd.concat([x_test, y_test], axis=1, sort=False)

    return train, test
```

```python
def to_csv(df: pd.DataFrame, saving_path: str):
    if not os.path.isfile(saving_path):
        df.to_csv(saving_path, index=False)
    else:
        df.to_csv(saving_path, index=False, mode='a', header=False)
```

```python
def preprocessing_all(le: LabelEncoder, chunksize=1000000):
    for chunk in pd.read_csv(FILE_PATH, skipinitialspace=True, chunksize=chunksize):
        train, test = split_train_test(process(chunk, le))
        to_csv(train, os.path.join(PROCESSED_DIR_PATH, "train_MachineLearningCVE.csv"))
        to_csv(test, os.path.join(PROCESSED_DIR_PATH, "test_MachineLearningCVE.csv"))
```

```python
label_encoder = label_encoding()
```

```python
preprocessing_all(label_encoder, 2500000)
```
